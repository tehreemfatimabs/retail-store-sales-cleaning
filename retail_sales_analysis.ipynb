# Step 1 Importing libraries

```python
import pandas as pd
import numpy as np
```

# Step 2 Load Dataset

```python
df = pd.read_csv("retail_store_sales.csv")
```

# Step 3 Basic EDA (Exploratory Data Analysis)

```python
print("First Five rows")
print(df.head())
print("\nLast Five rows")
print(df.tail())
("Summary Statistics")
df.describe()
```

## Step 3 Identify Missing and Duplicate Records

```python
print("\n Missing Values")
df.info()
print(df.isnull().sum())

print("\nDuplicate Records")
df.duplicated().sum()
```

# Step 4 Handling Missing Data

```python
# Item
df['Item'] = df['Item'].fillna('Unknown')

# Discount Applied
df['Discount Applied'] = df['Discount Applied'].fillna('Unknown')

# Numerical columns
df['Price Per Unit'] = df['Price Per Unit'].fillna(
    df['Price Per Unit'].median()
)

df['Quantity'] = df['Quantity'].fillna(
    df['Quantity'].median()
)

df['Total Spent'] = df['Total Spent'].fillna(
    df['Total Spent'].median()
)
```

```python
print(df.isnull().sum())
```

## Step 5 Inconsistent Date Format

```python
df['Transaction Date'] = pd.to_datetime(
    df['Transaction Date'],
    errors='coerce'
)
```

```python
df.info()
```

## Step 6 Saved Clean Dataset

```python
df.to_csv('retail_store_sales_cleaned.csv', index=False)
```

### Why This Approach Was Used

The following approach was used to prepare the retail sales dataset for reliable analysis:

- **Missing categorical values (`Item` and `Discount Applied`)** were replaced with `"Unknown"` instead of removing the records. This preserves the transaction records while clearly indicating that the information was unavailable.

- **Missing numerical values (`Price Per Unit` and `Quantity`)** were replaced using the **median**. The median is less affected by extreme values and is therefore suitable for retail data that may contain unusually high or low prices or quantities.

- **Missing `Total Spent` values** were first calculated using `Price Per Unit × Quantity` where possible. Remaining missing values were filled using the median to avoid losing transaction records.

- **Duplicate records** were checked and none were found, so no rows were removed for duplication.

- **Transaction Date** was converted from text to a proper datetime format to make the dataset suitable for date-based analysis.

This approach was chosen to **retain as much original data as possible while maintaining consistency and reliability for further analysis**.
